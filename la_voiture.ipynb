```python
class Voiture:
    def __init__(self, model, color, year):
        self.model = model
        self.color = color
        self.year = year
        self.km = 0

    def decrire_voituire(self):
        return f" {self.model} {self.color} {self.year}"

    def lire_km(self):
        return f"Le compteur de la voiture est à {self.km} km"
```
